### Types of errors in File Handling:

```python
# file does not exist at the specified path
try:
    with open("abc.txt", "r") as f:
        content = f.read()
except FileNotFoundError as e:
    print( e)
```

```python
# Invalid mode
try:
    with open("my_file.txt", "abc") as f:
        pass
except ValueError as e:
    print(e)
```

```python
# file already exists
try:
    with open("sample_files/sample_file.txt", "x") as f:
        content = f.read()
except Exception as e:
    print(e)
```

```python
# trying to open a directory as a file
try:
    with open("sample_files", "w") as f:
        f.write("Attempting to write in a directory.")
except PermissionError as e:
    print(e)
```

```python
# trying to open a directory as a file is also an IOError/OSError
try:
    with open("sample_files/sample_dir", "r") as f:
        content = f.read()
except (OSError, IsADirectoryError) as e:
    print(e)
```
