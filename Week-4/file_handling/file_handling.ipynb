### Modes in file handling:

[r/w/a/x] [+] [b/t]


'r'                 open for reading (default)

'w'                 open for writing, truncating the file first

'x'                 create a new file and open it for writing

'a'                 open for writing, appending to the end of the file if it exists

'b'                 binary mode

't'                 text mode (default)

'+'                 open a disk file for updating (reading and writing)

'U'                 universal newline mode (deprecated)

```python
with open("sample_files/sample_file.txt", "r") as file:      #using read mode
    for line in file:                           #iterating line by line
        print(line.strip())

#or,

file = open("sample_files/sample_file.txt", 'r')
try:
    for line in file:                           #iterating line by line
        print(line.strip())
finally:
    file.close()
```

```python
with open("sample_files/sample_file.txt", "r") as file:
    print(file.read())                          #using read() method
```

```python
with open("sample_files/sample_file.txt", "r") as file:
    print(file.read(3))
    print(file.read(10))
    print(file.read(100))
    print(file.read(100))

    # using seek(offset, whence) method to change pointer position
    file.seek(0)
    print(file.tell())
    print(file.read(1))
    print(file.tell())

    file.seek(5)
    print(file.read())

    # file.seek(-4,2)
    # print(file.read(5))

    print(file.tell())
```
