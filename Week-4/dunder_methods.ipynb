### Dunder methods used so far:

#### Classes:

\_\_new\_\_(): creates new instance of a class

\_\_init\_\_(): initializes attributes of the instance

\_\_add\_\_(): defines the behavior of the + operator

\_\_str\_\_(): defines string representation of an object

```python
class CustomObject:
    def __new__(cls, value):

        instance = super().__new__(cls) # Creating new instance
        instance.value = value + 10
        return instance

    def __init__(self, new_value):
        print('init called')

    def __add__(self, other):
        return CustomObject(self.value + other.value)

    def __str__(self):
        return f'object with value: {self.value} created!'

obj1 = CustomObject(5)
print(obj1)
obj2 = CustomObject(10)
print(obj1.value)

print(obj1 + obj2)
```

#### Iterators:

\_\_iter\_\_(): initializes and returns the iterator object

\_\_next\_\_(): retrieves the next item in the sequence

```python
class Even:
    def __iter__(self):
        self.n = 2
        return self        # must return the iterator object itself

    def __next__(self):
        x = self.n
        self.n += 2
        return x

even = Even()
it = iter(even)
print(next(it))
```

#### Context Manager:

'with' keyword is used for file resource management (ensuring they are properly acquired and released).
Using with open() automatically closes the file when the task is completed, even if an error occurs.

With statement relies on context managers, which manage resource allocation and deallocation using two special methods:

\_\_enter\_\_(): Acquires the resource and returns it

\_\_exit\_\_(): Releases(cleans up) the resource when the block exits

With statement is also used in managing database connections.

```python
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None
        print('init method called')

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        print('enter method called')
        return self.file

    def __exit__(self, exc_type, exc_value, exc_traceback):
        self.file.close()
        print('exit method called')

with FileManager('file_handling/sample_files/sample_file2.txt', 'a') as f:
    print('with block')
    f.write('Test')

print(f.closed)
```
